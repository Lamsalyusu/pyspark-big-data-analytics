```python
# printing fibonacci sequence using function
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        print(a, end=' ')
        a, b = b, a + b
# user input for number of terms in the Fibonacci sequence
num_terms = int(input("Enter the number of terms in the Fibonacci sequence: "))
fibonacci(num_terms)


```

```python
# Factorial of a number using fucntion
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)
# number to calculate factorial
num = int(input("Enter a number to calculate its factorial: "))
result = factorial(num)
print(f"The factorial of {num} is: {result}")
```

```python
# printing right angle triangle pattern using for loop
def right_angle_triangle(n):
    for i in range(1, n + 1):
        print('* ' * i)
# user input 
num_rows = int(input("Enter the number of rows for the right angle triangle: "))    
right_angle_triangle(num_rows)
# print()
```

```python
# making a simple calculator using functions
def add(x, y):
    return x + y    
def subtract(x, y):
    return x - y
def multiply(x, y):
    return x * y
def divide(x, y):
    if y != 0:
        return x / y
    else:
        return "Error! Division by zero."
# user input for numbers and operation
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operation = input("Enter operation (+, -, *, /): ")
if operation == '+':
    result =add(num1, num2)
    print ("The result of the addition is: ", result)
elif operation == '-':
    result = subtract(num1, num2)
    print ("The result of the subtraction is: ", result)
elif operation == '*':
    result = multiply(num1, num2)
    print ("The result of the multiplication is: ", result)
elif operation == '/':
    result = divide(num1, num2)
    print ("The result of the division is: ", result)
else:
    print("Invalid operation! Please enter one of +, -, *, /.")
    
```

```python
# removing the first duplicate element from a list
def remove_first_duplicate(lst):
    seen = set()
    for i in range(len(lst)):
        if lst[i] in seen:
            return lst[:i] + lst[i+1:]
        seen.add(lst[i])
    return lst
# user input for list elements
input_list = input("Enter a list of elements separated by spaces: ").split()
result_list = remove_first_duplicate(input_list)
print("List after removing the first duplicate element: ", result_list)
```

```python
# Create a new dictionary by extracting only specific keys from a given dictionary — extract "name" and "salary" from sample_dict
sample_dict = {
    "name": "John Doe",
    "age": 30,
    "salary": 50000,
    "department": "HR"
}
def extract_specific_keys(d, keys):
    return {key: d[key] for key in keys if key in d}
keys_to_extract = ["name", "salary"]    
new_dict = extract_specific_keys(sample_dict, keys_to_extract)
print("New dictionary with specific keys: ", new_dict)
```
